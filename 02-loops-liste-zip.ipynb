```python
# Stato attuale del robot (letture simulate)
nome_robot = "UR5e-01"
temperatura = 42.5
distanza_cm = 5
batteria_percento = 10

# Soglie di sicurezza — valori tipici di un sistema di monitoring reale
soglia_batteria_bassa = 20
soglia_distanza_critica = 15
soglia_temperatura_alta = 60.0

# Ogni condizione salvata in una variabile bool separata: più leggibile e testabile
batteria_bassa = batteria_percento < soglia_batteria_bassa
distanza_critica = distanza_cm < soglia_distanza_critica
temperatura_alta = temperatura > soglia_temperatura_alta

if batteria_bassa and distanza_critica:
    stato = "DOPPIO ALLARME"
elif distanza_critica or temperatura_alta:
    stato = "CRITICO — arresto immediato"
elif batteria_bassa:
    stato = "ATTENZIONE — batteria in esaurimento"
else:
    stato = "OK — operativo"

print(f"Robot {nome_robot}: temperatura {temperatura}°C, batteria {batteria_percento}%, distanza {distanza_cm}cm, stato={stato}")
```

```python
batteria_percento_letture = [45, 30, 22, 32, 26, 53, 40]
soglia_distanza_critica = 20

conteggio_critici = 0   # "accumulatore" — parte da zero, cresce durante il ciclo

for lettura in batteria_percento_letture:
    if lettura < soglia_distanza_critica:
        print(f"{lettura}cm -> ATTENZIONE")
        conteggio_critici += 1
    else:
        print(f"{lettura}cm -> OK")

print(f"Totale letture critiche: {conteggio_critici} su {len(batteria_percento_letture)}")
```

```python
distanza_letture = [45, 30, 22, 14, 9, 18, 40]
batteria_letture = [80, 75, 15, 10, 50, 18, 90]

soglia_distanza_critica = 15
soglia_batteria_bassa = 20

for distanza, batteria in zip(distanza_letture, batteria_letture):
    distanza_critica = distanza < soglia_distanza_critica
    batteria_bassa = batteria < soglia_batteria_bassa

    if batteria_bassa and distanza_critica:
        stato = "DOPPIO ALLARME"
    elif distanza_critica:
        stato = "CRITICO — arresto immediato"
    elif batteria_bassa:
        stato = "ATTENZIONE — batteria in esaurimento"
    else:
        stato = "OK — operativo"

    print(f"distanza={distanza}cm, batteria={batteria}% -> {stato}")
```
